# Member A — distributions of the supplied smoking data

**Question.** What does the sample look like before studying the relationship between smoking and survival? These are unadjusted, one-variable descriptions, not causal estimates. The displayed sample and variables were checked against D's data-preparation PR #13: it retains all rows, drops only the unused index column, and adds binary smoking and survival indicators. Salary remains missing for 700 people; figures use the observed values without imputation. Start Jupyter in the repository root and use the `smoke` kernel.

```python
from IPython.display import display
from member_a_distributions import (
    load_data, plot_numeric_distributions, plot_categorical_distributions, save_figures
)

data = load_data()
print(f"Rows: {len(data):,}")
display(data[["age", "salary"]].describe())
display(data[["age", "salary", "smoker", "outcome", "gender"]].isna().sum().rename("missing"))
for column in ("smoker", "outcome", "gender"):
    display(data[column].value_counts(dropna=False).rename(column))
```

```python
numeric_figure = plot_numeric_distributions(data)
display(numeric_figure)
categorical_figure = plot_categorical_distributions(data)
display(categorical_figure)
save_figures(data)
```

## Interpretation

The supplied dataset has 1,314 people. Age ranges from 18 to 84 years, with a median of 46; its distribution has a modest right skew (sample skewness about 0.20). Among the 614 observed salaries, the median is 48,692 and the distribution has little overall skew (about 0.07); the salary units are not documented in the CSV. Both variables have no observations beyond the conventional 1.5-IQR fences, although that rule alone does not establish whether values are valid.

There are 582 smokers (44.3%) and 732 non-smokers (55.7%). The recorded outcomes are 945 alive (71.9%) and 369 dead (28.1%); gender is 693 male (52.7%) and 621 female (47.3%). These describe marginal sample composition, not survival rates within smoking groups.

Salary is missing for 700 people (53.3%); the other four displayed variables have no missing values. The salary histogram describes only the observed 46.7%, so it may not represent the full sample if missingness is selective. Uneven outcome counts and the broad age range matter when interpreting later comparisons. The graphs do not establish a smoking effect or explain the apparent survival difference.
